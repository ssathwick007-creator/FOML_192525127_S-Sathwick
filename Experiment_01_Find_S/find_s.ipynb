1) FDD-S Algorithm (Find-S Algorithm)

Objective:
Implement and demonstrate the Find-S Algorithm to find the most specific hypothesis from a given training dataset.

```python
import pandas as pd
```

```python
data=pd.read_csv("enjoysport.csv")
data
```

```python
concepts = data.iloc[:, :-1].values
target = data.iloc[:, -1].values

print("Training Examples:")
print(concepts)

print("\nTarget:")
print(target)
```

```python
hypothesis = concepts[0].copy()

print("Initial Hypothesis:")
print(hypothesis)
```

```python
for i in range(len(concepts)):
    if target[i] == "Yes":
        for j in range(len(hypothesis)):
            if hypothesis[j] != concepts[i][j]:
                hypothesis[j] = "?"

print("Final Hypothesis:")
print(hypothesis)
```

Algorithm:
1) Starts with the first positive example.
2) Compares it with every other positive training example.
3) Whenever an attribute differs, it replaces it with ?.
4) Negative examples are ignored.
5) The final hypothesis is the most specific hypothesis that covers all positive examples.

Result:
The Find-S algorithm was successfully implemented using Python. It learned the most specific hypothesis that correctly classifies all positive training examples by generalizing only when necessary.
